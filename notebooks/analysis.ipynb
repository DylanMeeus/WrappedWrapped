# WrappedWrapped analysis

Use this notebook to explore the cached Spotify snapshots in `data/raw/`.

```python
import sys
from pathlib import Path

sys.path.append(str(Path('..').resolve()))

from src.analysis import build_overlap, load_snapshots

snapshots = load_snapshots()
print(f"Loaded {len(snapshots)} snapshots")

track_overlaps, artist_overlaps = build_overlap(snapshots)
print(f"Tracks appearing in multiple years: {len(track_overlaps)}")
print(f"Artists appearing in multiple years: {len(artist_overlaps)}")

print("\nTop overlapping tracks:")
for entry in track_overlaps[:10]:
    years = ', '.join(str(year) for year in entry['years'])
    artists = ', '.join(entry['artists']) or 'Unknown artist'
    print(f"- {entry['name']} — {artists} ({years})")
```
